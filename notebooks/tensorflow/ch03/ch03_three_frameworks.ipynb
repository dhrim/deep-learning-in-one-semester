```python
import numpy as np
import matplotlib.pyplot as plt

import dlbook
from dlbook import data, metrics, plot

dlbook.set_seed(42)
plot.use_korean()
print(dlbook.versions())
```

## 3.1 데이터 — 세 판이 글자까지 같은 셀

`dlbook.data` 는 **numpy 배열**을 돌려줍니다. tf.Tensor도 torch.Tensor도
아닙니다. 그래서 이 셀이 세 판에서 같을 수 있습니다.

```python
x, y = data.apples(n=400, seed=42)
s = data.split(x, y, val_ratio=0.2, test_ratio=0.2, seed=42)
print(s.summary())
```

## 3.2 모델 정의 — 여기부터 판마다 다릅니다

만드는 것은 같습니다. **가중치 2개와 편향 1개.** 2장 〈그림 2-7〉의 그 구조입니다.

```python
import tensorflow as tf

dlbook.set_seed(42)   # 프레임워크를 불러온 뒤 한 번 더

# 학습 루프를 직접 쓰므로 변수도 직접 만든다.
W = tf.Variable(tf.random.normal([2, 1], stddev=0.5), name="W")
b = tf.Variable(tf.zeros([1]), name="b")

def forward(inputs):
    return tf.sigmoid(tf.matmul(tf.cast(inputs, tf.float32), W) + b)

optimizer = tf.keras.optimizers.SGD(learning_rate=0.1)
loss_fn = tf.keras.losses.BinaryCrossentropy()
print("학습할 값:", W.shape.as_list(), b.shape.as_list())
```

## 3.3 학습 — 여기도 판마다 다릅니다

무엇을 하는지 보십시오. 이름만 다르고 하는 일이 같습니다.

1. 예측한다
2. 정답과 견주어 손실을 구한다
3. **경사를 구한다** (2장 §2.11의 자동 미분)
4. **그 방향으로 가중치를 옮긴다** (2장 §2.10의 경사하강)

```python
train_ds = (tf.data.Dataset
            .from_tensor_slices((s.x_train, s.y_train.astype("float32")))
            .shuffle(512, seed=42).batch(16))

@tf.function
def train_step(xb, yb):
    with tf.GradientTape() as tape:                 # ③ 계산 과정을 기록한다
        loss = loss_fn(tf.reshape(yb, (-1, 1)), forward(xb))   # ② 손실
    grads = tape.gradient(loss, [W, b])             # ③ 경사를 구하고
    optimizer.apply_gradients(zip(grads, [W, b]))   # ④ 그 방향으로 내려간다
    return loss

history = {"loss": [], "val_loss": []}
for _ in range(dlbook.smoke.epochs(60)):
    losses = [float(train_step(xb, yb)) for xb, yb in train_ds]
    history["loss"].append(float(np.mean(losses)))
    history["val_loss"].append(float(loss_fn(
        tf.reshape(s.y_val.astype("float32"), (-1, 1)), forward(s.x_val))))

def fw_predict(g):
    return (forward(g).numpy().reshape(-1) > 0.5).astype("int64")
```

## 3.4 결과 — 세 판이 같은 숫자를 냅니다

같은 데이터, 같은 구조, 같은 학습률, 같은 epoch입니다.
그러니 같은 답이 나오는 것이 당연합니다.

```python
acc = metrics.accuracy(s.y_test, fw_predict(s.x_test))
dlbook.record("ch03_apple_test_acc", acc)

plot.decision_boundary(fw_predict, s.x_test, s.y_test, title="찾은 기준선")
plt.show()

plot.loss_curve(history)
plt.show()
```
